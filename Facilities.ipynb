```python
import pandas as pd

# Load the full national dataset (CSV version)
gdf = pd.read_csv('nga_health_facilities_v2_0_FULL.csv')

# Size of the ORIGINAL dataset
print(f"Original dataset -> Rows: {len(gdf)}, Columns: {len(gdf.columns)}")

# SELECTION - filter rows to Oyo State only
oyo = gdf[gdf['state'] == 'Oyo']

# PROJECTION - pick only the columns you need
columns_needed = ['nhfr_facility_code', 'facility_name', 'lga', 'ward',
                  'ownership', 'ownership_type', 'facility_level',
                  'facility_level_option', 'latitude', 'longitude']
oyo_selected = oyo[columns_needed]

# Save your own clean output
oyo_selected.to_csv('oyo_health_facilities.csv', index=False)

# Size of the extracted dataset
print(f"Oyo extract      -> Rows: {len(oyo_selected)}, Columns: {len(oyo_selected.columns)}")

# Optional: how much of the original the extract represents
print(f"Oyo is {len(oyo_selected) / len(gdf) * 100:.2f}% of the national rows")
```
